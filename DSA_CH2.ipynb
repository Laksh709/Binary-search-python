```python
class User :
    def __init__(self,username,name,email):
        self.username = username 
        self.name = name 
        self.email = email

    def introduce_yourself(self,guest_name):
        print("Hi {}! , my name is  {} and contact me at {}." .format(guest_name, self.name,self.email))
```

```python
user3 = User('Jane123' , 'Jane Doe' , 'Jane@345')
```

```python
user3.introduce_yourself('David')
```
