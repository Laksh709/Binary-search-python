```python
def locate_card(cards,query):
    lo,hi = 0,len(cards)-1
    while lo<=hi:
        mid = (low+hi)//2
        mid_number = cards[mid]
        print('low:', lo , 'high:', hi,'mid:',mid , 'mid_number:',mid_number)

        if mid_number == query:
            return mid
        elif mid_number< query:
            hi = mid - 1
            
        elif mid_number> query:
            lo = mid + 1
        

    return -1        

        
        
        
```
